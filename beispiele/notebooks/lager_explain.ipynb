# Bestand je Lager
Der Plan von `explain()` steht in der Ausgabe der Zelle (Spark Schicht 2).

```python
bestand = spark.read.parquet("daten/bestand")
artikel = spark.read.parquet("daten/artikel")
je_lager = bestand.join(artikel, "nr").groupBy("lager").count()
```

```python
je_lager.explain()
```

```python
je_lager.show()
```
